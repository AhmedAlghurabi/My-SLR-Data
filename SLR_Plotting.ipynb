```python
import plotly.graph_objects as go
import os

# Data for fluid medium
title = "Fluid Medium"
file_name = "fig_distribution_of_fluid_medium.pdf"
labels = ["Air", "Water", "Gas", "Oil", "Blackwater", "Compressed air", "Water-PVA", "Methane", "Propellant gas", "Air-water", "Water-oil", "Gas-water", "Air-water-oil", "Oil-water"]
sizes = [36, 3, 18, 6, 1, 1, 1, 1, 1, 12, 2, 4, 1, 1]

# Preferred templates
templs = ['ggplot2', 'seaborn', 'simple_white', 'plotly', 'plotly_white', 'plotly_dark', 'presentation', 'xgridoff', 'ygridoff', 'gridon', 'none']

# Create a folder named Figures if it doesn't exist
if not os.path.exists('Figures'):
    os.makedirs('Figures')

# Create a single donut chart
fig = go.Figure(
    data=[go.Pie(
        labels=labels,  # Specify labels for each segment of the pie chart
        values=sizes,  # Specify values for each segment, determining their relative sizes
        hole=.3,  # Set the "hole" size for the donut chart (0.3 makes it a donut chart)
        opacity=1,  # Set opacity to make the chart fully opaque
        sort=True,  # Enable sorting of segments by value
        textinfo='text',  # Specify custom text to display inside each segment
        text=[f"<b>{label}</b><br>n={size}, {size / sum(sizes):.0%}" for label, size in zip(labels, sizes)],
        textposition='inside',  # Position the custom text inside the chart segments
        insidetextfont=dict(size=12, color="white"),  # Set font size and color for text inside segments
        marker=dict(line=dict(color='grey', width=0.1))  # Add a grey border around each segment with minimal width
    )]
)

# Update layout of the chart
fig.update_layout(
    showlegend=False,  # Hide the legend for cleaner appearance
    autosize=False,  # Disable autosizing to set custom dimensions
    width=600,  # Set width of the figure
    height=600,  # Set height of the figure
    template=templs[3],  # Apply a predefined template for styling the chart
    margin=dict(l=10, r=10, t=30, b=10),  # Reduce the margins around the figure
)

# Add a centered annotation (e.g., title) in the middle of the chart
fig.add_annotation(
    text=title,  # Text for the annotation (e.g., chart title)
    x=0.5, y=0.5,  # Center the annotation at the middle of the chart
    showarrow=False,  # Disable arrow for the annotation
    font=dict(size=16, color="black"),  # Set font size and color for the annotation
    xanchor='center', yanchor='middle'  # Anchor the text to the center
)

# Show the plot in an interactive window


# Save the figure to an image file in the Figures folder
fig.write_image(os.path.join('Figures', file_name))
fig.show()
```

```python

import plotly.graph_objects as go
import pandas as pd
import os

# Data
data = {
    "Database Screening": ["Wiley", "Scopus", "ScienceDirect", "Web of Science", "OnePetro", "MDPI"],
    "# Studies Identified": [20, 710, 750, 792, 154, 335],
    "# Studies Passed Duplicates Discard": [20, 571, 689, 757, 150, 334],
    "# Studies Passed Screening 1st round": [2, 61, 153, 34, 28, 15],
    "# Studies Passed Screening 2nd round": [0, 18, 114, 19, 19, 5],
    "# Studies Passed Screening Final round": [0, 18, 114, 19, 19, 5],
    "# Studies Passed Eligibility Assessment": [0, 5, 65, 6, 5, 3],
    "# Studies Included": [0, 5, 65, 6, 5, 3]
}

# Convert data to DataFrame
df = pd.DataFrame(data)

# Preferred templates
templs = ['ggplot2', 'seaborn', 'simple_white', 'plotly', 'plotly_white', 'plotly_dark', 'presentation', 'xgridoff', 'ygridoff', 'gridon', 'none']

# Create a folder named Figures if it doesn't exist
if not os.path.exists('Figures'):
    os.makedirs('Figures')

# Plot each column in a separate donut chart and save as PDF
for column in df.columns[1:]:
    fig = go.Figure(data=[go.Pie(
        labels=df["Database Screening"],
        values=df[column],
        hole=0.3,
        opacity=1,
        sort=True,
        textinfo='text',
        text=[f"<b>{label}</b><br>n={size}, {size / sum(df[column]):.0%}" for label, size in zip(df["Database Screening"], df[column])],
        textposition='inside',
        insidetextfont=dict(size=12, color="white"),
        marker=dict(line=dict(color='grey', width=0.1))
    )])

    # Update layout with no title
    fig.update_layout(
        showlegend=False,
        autosize=False,
        width=600,
        height=600,
        template=templs[3],
        margin=dict(l=10, r=10, t=30, b=10),
    )

    # Add centered annotation in the middle of the chart
    fig.add_annotation(
        text="Database",
        x=0.5, y=0.5,
        showarrow=False,
        font=dict(size=16, color="black"),
        xanchor='center', yanchor='middle'
    )

    fig.show()

    file_name = f"fig_{column.replace(' ', '_').replace('#', '').replace('/', '_').lower()}.pdf"
    fig.write_image(os.path.join('Figures', file_name))

print("Plots saved successfully in the Figures folder.")

```

```python
import plotly.graph_objects as go
import os

# Data
data = {
    "Score": [0, 0.5, 1, 1.5, 2, 2.5, 3, 3.5, 4],
    "# of papers": [0, 0, 1, 9, 17, 32, 32, 41, 43]
}

# Create a folder named Figures if it doesn't exist
if not os.path.exists('Figures'):
    os.makedirs('Figures')

# Create a bar chart
fig = go.Figure()

# Add bars for each score
colors = ['blue'] * len(data["Score"])
max_indices = [i for i, val in enumerate(data["# of papers"]) if val in sorted(data["# of papers"])[-2:]]
for i, val in enumerate(data["# of papers"]):
    if i in max_indices:
        colors[i] = 'red'

fig.add_trace(go.Bar(
    x=data["Score"],
    y=data["# of papers"],
    marker_color=colors,
    text=data["# of papers"],
    textposition='auto'
))

# Update layout
fig.update_layout(
    title="Number of Papers by Score",
    xaxis_title="Score",
    yaxis_title="# of Papers",
    showlegend=False,
    autosize=False,
    width=800,
    height=600,
    template='plotly_white',
    margin=dict(l=20, r=20, t=30, b=20),
)

# Show the plot
fig.show()

# Save the figure to an image file in the Figures folder
file_name = "fig_number_of_papers_by_score.pdf"
# fig.write_image(os.path.join('Figures', file_name))

print("Plot saved successfully in the Figures folder.")
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import os

# Data provided
data = {
    "Ratings": [0, 1, 2],
    "# of papers after 1st round of screening": [2228, 293, 0],
    "# of papers after 2nd round of screening": [2346, 175, 0], # Note: Assuming 0 for rating 2 in 2nd round based on other data
    "# of papers after Final round of screening": [2228, 118, 175],
    "Total Screened": [2521, 2521, 2521] # This row seems to be a total, not per rating, so handle separately
}

# Preferred templates (from your style)
templs = ['ggplot2', 'seaborn', 'simple_white', 'plotly', 'plotly_white', 'plotly_dark', 'presentation', 'xgridoff', 'ygridoff', 'gridon', 'none']
template_to_use = templs[3] # Using 'plotly' template as before

# Create a folder named Figures if it doesn't exist
if not os.path.exists('Figures'):
    os.makedirs('Figures')

# --- 1. Bar chart for the number of publication per each rate (using Final round data) ---
bar_labels = [f"Rating {r}" for r in data["Ratings"]]
bar_values = data["# of papers after Final round of screening"]
total_final_screened = sum(bar_values) # Calculate sum from the ratings

# Filter out ratings with 0 papers for better bar chart visualization if desired, or keep all
# For this bar chart, let's keep all and show 0 if present.

fig_bar = go.Figure(
    data=[go.Bar(
        x=bar_labels,
        y=bar_values,
        marker_color=['skyblue', 'lightcoral', 'lightgreen'], # Example colors
        text=[f"n={v}" for v in bar_values], # Text on top of bars
        textposition='outside',
        textfont=dict(size=12, color="black")
    )]
)

fig_bar.update_layout(
    # title_text="Number of Publications per Rating (After Final Round)",
    xaxis_title="Rating",
    yaxis_title="# of Papers",
    template=template_to_use,
    width=700,
    height=500,
    margin=dict(l=50, r=50, t=50, b=50),
    # Add annotation for total if needed, similar to the donut chart style
    annotations=[
        go.layout.Annotation(
            text=f"Total Papers (Final Screened): {total_final_screened}",
            x=0.5, y=1.05, xref='paper', yref='paper', showarrow=False,
            font=dict(size=14, color="black"),
            xanchor='center', yanchor='bottom'
        )
    ]
)

fig_bar.show()
# fig_bar.write_image(os.path.join('Figures', "fig_bar_papers_per_rating_final.pdf"))
print("Generated fig_bar_papers_per_rating_final.pdf")
```
